# Hyperliquid: who trades it, who wins, and whether they can be found

**Markout analysis of the full Hyperliquid perp tape (Aug 2025 – Jun 2026), and the search for informed flow.**

*Scope: majors analyzed individually are BTC, ETH, SOL, HYPE, and SPX. 202 standard perp
markets over the 11 months; builder-deployed micro-markets (opaque #-numbered identifiers)
are excluded by data-quality floors. No standard ticker corresponds to SpaceX — a
builder-deployed SpaceX market cannot be ruled out, as those use opaque identifiers.*

**TL;DR.** The average eligible Hyperliquid taker loses ~16bp per entry over the following
24h (median wallet: −13bp). Ranking wallets by markout PnL — the original design — does
**not** identify traders whose edge persists (pre-registered out-of-sample test: fail). But
informed traders demonstrably exist: 57 wallets carry certification scores where a
dependence-preserving null expects ~15 — implying **≈40 genuinely informed wallets**, which
entity analysis dedupes to **43 independent information sources** (so ~28 more sources than
luck explains). They are identifiable as a basket (~74% genuine), not as certified
individuals at this data length. Their signature: legible in alt-coin records, weak in
majors records, and overwhelmingly **contrarian** (77% faders). July data provides the
out-of-sample confirmation.

Eleven months of complete fills tape: ~50M fills/month, both counterparties on every trade,

orders, ≥25 active days, 0.5–15 orders/day, ≥$500 mean order size — about 23k wallets. This
band is a *mechanical* choice (enough evidence to measure, slow enough to copy), not a

### 1.5 The baseline fact

**Markout** is this report's core measurement: after a wallet enters a position, how far does
price move in its favor over the next N hours, in basis points? Entries are priced at the close
of the first 5-minute bar *after* the fill (never the fill's own bar — pricing inside the
entry's own bar leaks the trader's information into the benchmark and once manufactured a fake
edge in an earlier project phase; appendix A).

The distribution below is each eligible wallet's average 24h markout across all its entries.
**The field is not neutral — the mean wallet loses ~16bp per entry (median: −13bp).** Copying
a random Hyperliquid trader is a reliably losing strategy; anything that claims to find skill
must first climb out of this hole:

## 2. Markout by market and horizon — the core assignment

Mean markout at nine horizons (1h → 168h) for each major market and the alt aggregate. Two
readings: the **level** says what the average entry earns; the **shape** distinguishes
information (drift that persists or grows) from impact (a spike that decays).


<img src="report_figs2/07_term_structure.png" width="980">


Alts earn a small positive pop right after entry (+1.9, +2.6, +0.7 bps at 1h, 2h, 4h) and then bleed as the horizon lengthens, reaching about −195bp by a week out. Majors sit near zero for the first day, but bleed on longer horizons too — SPX is the worst tail at about −368bp by 168h. No market pays you for simply holding after a random entry, which foreshadows §6's finding that majors *records* carry almost no selection signal.

## 3. Filtering: reasoning and consequences

Beyond the §1 activity band, three filters survived testing, all *mechanical*: a ≥1h median
position-hold floor (a copier's detection latency makes faster traders unharvestable),
exclusion of TWAP-slice and liquidation fills from scoring (they are position changes, not
decisions), and a maker-exit cap applied only in copy simulations (resting exits can't be
mirrored). Every **performance-shaped** filter tested — concentration caps, effective-bet
floors, win-breadth floors, maker exclusion — *failed* a matched-population certification:
each removed wallets that went on to perform as well or better than those kept.

The rule that emerged: **filters fix mechanics; statistics handle luck.**


Below, distribution comparisons between the full eligible field and the top-50 by 24h markout
PnL (all markets combined, train window) across six dimensions: total orders, active days,
orders/day, mean order size, months active, and average 24h markout. The selected wallets are
bigger and more active than the field they came from:


<img src="report_figs/12_cohort_dists.png" width="980">


The most instructive rejected filter: excluding maker-heavy wallets. Certification showed
transfer *rising* with maker share — when a habitually passive wallet crosses the spread and
pays to trade, that entry carries more information, not less. This inverted our prior
completely and becomes one of §6's structural findings:

## 5. Informed traders exist — the existence analysis

The right question is not "who made the most dollars" but "whose *record* is too consistent
to be luck." The instrument (final form, after two documented failed versions — appendix B):

1. Score every entry against **its own coin's** monthly average outcome (so riding a hot coin
   or a bull month earns nothing).
2. Merge same-coin-same-day entries into one **position** (a 40-fill sweep is one decision).
3. Per wallet, compute a **t-statistic**: mean position excess over its own noise, on raw
   uncapped outcomes — evidence grows with the number of genuinely independent positions.
4. Compare the whole field against a **direction-scrambled null**: each wallet's own record
   with week-blocks of directions flipped — preserving every correlation, overlap, and
   fat tail while destroying skill. No independence or normality assumptions anywhere.

blue is reality. The right tail separates:

## 6. Where information lives: three structural findings

**Copyability ≠ identifiability.** Majors positions are mechanically cheap to copy (deep
books, tight spreads — the binding costs in our follower model are fees, not market impact)
— but majors *records* are weak identifiers of skill: everyone's BTC PnL is dominated by the
same regime. Alt records are idiosyncratic and therefore legible. In exploratory tests
(pooled z ≈ +1.1, consistent across all folds tried; **July confirms or breaks this**),
ranking wallets by alt performance predicted even their *majors* trading next month, while
the reverse predicted nothing:


<img src="report_figs2/10_venue_matrix.png" width="980">


**The maker paradox** (§3): informed flow is disproportionately maker-heavy wallets —
studied here only through their taker entries. Their passive-side behavior is the largest
unexplored territory in this report.

**The field is anti-informed in aggregate** (mean −16bp/entry, §1.5). The dependence-honest
test finds the *negative* tail indistinguishable from chance (19 wallets beyond −4 vs ~15
expected) — so the field's losses are broad and structural, not concentrated in certifiably
anti-skilled individuals, and no individual "fade pool" survives certification.

### 7.2 Some of these are the same trader

Comparing all 57 pairwise — co-timed entries (same coin, same direction, within 10
minutes), exposure overlap, and daily PnL correlation — collapses them to **43
independent information sources**.

The matrix below shows every measured pairwise similarity, with wallets ordered so
entity groups sit together (green boxes: the 8-, 4-, 3-, 2-, and 2-wallet groups).
Bright blocks inside green boxes = wallets that demonstrably trade as one; the dark
field elsewhere = genuine independence. Every pixel is a measured number — there are
no layout choices in this figure:


<img src="report_figs2/13_entity_network.png" width="980">


On the **style axis**, the certified basket is decisively one-sided: **77% of the
certified wallets are faders** (field: 36%) — median momentum share 0.32 vs the field's
0.51, with a long tilt (0.71). The venue's statistically-certified informed flow is, to
first order, **dip-buying**: providing liquidity into 4-hour dislocations and collecting
forward drift. Two precision notes: the *certification* is measured on forward returns
against a month-constant benchmark (immune to the backward-looking-baseline artifact of
appendix B); the *style labels* are descriptive classifications from trailing price moves
and carry no skill claim by themselves.


<img src="report_figs2/18_style_map.png" width="1050">


The 8-wallet network is itself a finding: each member clears the certification bar, but
their evidence is largely **shared** (50–87% co-timed entries) — one underlying strategy
observed eight times, not eight independent confirmations. Whoever operates it is either one
entity fragmenting across wallets or a leader already being copied on-chain; following one
member captures the group's signal.

## 8. Limitations, and what runs next

This analysis embeds a theory of informed flow: regular, taker-side, wallet-sized,
medium-frequency. Four known blind spots, each now a declared follow-up experiment:
**passive-side information** (maker fills were captured but never scored), **episodic traders**
(crisis-only specialists fail the activity floors), **position reductions** (selling ahead of a
decline is information our event definition ignores), and **delisting censoring** (dead coins'
final losses vanish from scoring — flagged, unfixed, flatters alt results).

**Next steps** (specs frozen before July data existed): July out-of-sample confirmation of the
basket and the alt-selection signal; the maker→taker transition detector; the delisting fix; a
broad-universe existence pass; and a live paper harvester (built, audited, currently stopped)
running the 43-slot basket.
