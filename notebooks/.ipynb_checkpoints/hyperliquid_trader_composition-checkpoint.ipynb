# Hyperliquid: who trades it, who wins, and whether they can be found

**Markout analysis of the full Hyperliquid perp tape (Aug 2025 – Jun 2026), and the search for informed flow.**

*Scope note: majors analyzed individually are BTC, ETH, SOL, HYPE, and SPX (present in the tape,
included throughout). SpaceX does not appear in the perp fills archive under any ticker — likely a
builder-deployed market outside the standard perp set.*

**TL;DR.** The average Hyperliquid taker *loses* ~14bp per entry over the following 24h. Ranking
wallets by markout PnL — this project's original design — does **not** identify traders whose edge
persists (pre-registered out-of-sample test: fail). But informed traders demonstrably exist: a
statistically rigorous screen finds **~40 independent informed information sources** among 22,000
eligible wallets — identifiable as a basket (57 wallets, ~74% genuine), never as certified
individuals at this data length. Their signature: legible in **alt-coin records**, invisible in
majors records, disproportionately **maker-heavy** wallets. Live confirmation runs on July data.

### 1.5 The single most important fact about the field

Score every wallet's entries by **24h markout** (signed move from the first 5-minute bar after
the fill to 24h later — never the fill's own bar; that distinction once manufactured a fake edge
and is now law here). The average wallet is *paying* the market to trade:

## 2. Markout by market — the assignment, executed

Mean markout at nine horizons (1h → 168h) per market. Two readings matter: the **level** (does
the average entry earn anything) and the **shape** (a plateau suggests information; a spike that
decays suggests impact).


<img src="report_figs2/07_term_structure.png" width="980">


Ranking traders by markout PnL at each horizon was the original deliverable. The rankings
agree strongly across mid horizons (8h–72h) and diverge at the extremes — a 1h ranking and a
weekly ranking select different traders:

### 2.4 Why a ranking is not an answer

The top-50 by first-half markout PnL, watched over the next six months: the cloud regresses.
Big past PnL is mostly big books × luck. This motivates everything in §3–5:

## 3. Filtering: from 470k wallets to a defensible cohort

Filters that survived all tests are **mechanical** (can this wallet's behavior be copied at all?):
the activity band, a ≥1h median-hold floor (copy latency physics), excluding TWAP/liquidation
fills from scoring (they are position changes, not decisions), and a maker-exit cap on the
follower side only. Every **performance-shaped** filter we tried — concentration caps,
effective-bets floors, win-breadth floors, maker exclusion — failed a matched-population test:
each removed *good* wallets. Rule that emerged: **filters fix mechanics; statistics fix luck.**


<img src="report_figs/12_cohort_dists.png" width="980">


The most instructive rejection: excluding maker-heavy wallets *hurt*. Transfer rises
monotonically with maker share — when a habitually passive wallet pays the spread, that entry
carries more information, not less:


<img src="report_figs2/09_maker_bands.png" width="980">

## 4. Does markout-PnL ranking find informed traders? (the sealed test)

Pre-registered design: rank on 8 train months, test once on a sealed Apr–Jun 2026 window,
decision bars declared in advance, 7 rounds of adversarial code audit. Verdict:

| test | result | bar | verdict |
|---|---|---|---|
| markout cell | 84.5th pct vs placebos | 97.5th | **FAIL** |
| follower simulation (with costs) | Sharpe −0.03 | CI > 0 | **FAIL** |
| month-by-month walk-forward | 3/5 folds, z +0.37 | 4/5, z > 1.64 | **FAIL** |

The selected wallets' own PnL went negative in the test quarter:


<img src="report_figs2/08_verdict_folds.png" width="980">


**Bounded conclusion:** performance ranking did not identify copyable traders within this
medium-frequency taker cohort. It does *not* establish that Hyperliquid lacks informed flow —
which is exactly the question §5 answers.


---
## 5. Informed traders exist — the existence analysis

The correct instrument (arrived at after two under- or over-powered attempts, both documented in
the appendix): score each wallet's entries against **its own coin's** monthly average, cluster
same-coin-same-day entries into one decision, compute a per-wallet t-statistic on raw (uncapped)
outcomes, and compare the field against a **direction-scrambled null** — each wallet's own record
with week-blocks of directions flipped, which preserves every correlation and overlap while
destroying skill. No independence assumptions anywhere.


<img src="report_figs2/06_flipnull.png" width="980">


At t>4: **57 wallets observed where luck-with-all-correlations expects 15** — and the
anti-skill side sits exactly at its null (19 vs 15). Entity clustering (co-timed entries, exposure
overlap, PnL correlation) collapses the 57 to **43 independent information sources** — including
one 8-wallet network trading as a single mind (50–87% of entries co-timed). The informed
population is broad, real, and roughly ¾ of the basket by construction.

**What 11 months cannot do:** certify any *individual* wallet (best achievable false-discovery
rate ≈ 26%). The product this data supports is the **deduplicated 43-slot basket**, refreshed
monthly — cohort-level out-of-sample transfer ≈ +1σ gross (~+12bp/entry, positive 5 of 8 months),
net-of-cost copyability pending the follower simulation, July data, and the live paper run.


---
## 6. Where information lives: three structural findings

**Copyability ≠ identifiability.** Majors positions transfer mechanically at low cost (~2.5bp/side
breakeven) — but majors *records* cannot identify skill: everyone's BTC PnL is the same regime.
Four independent instruments agree. Alt records are idiosyncratic and therefore legible — and alt
skill *predicts* majors performance (the reverse predicts nothing):


<img src="report_figs2/10_venue_matrix.png" width="980">


**The maker paradox** (fig 3.2 above): the venue's informed flow is disproportionately
maker-heavy wallets — studied here only through their taker entries. Their passive-side behavior
is the largest unexplored territory in this report.

**The field is anti-informed** (−14bp/entry): fading the aggregate taker field is itself a
candidate signal, and 257 wallets are certified *anti-skilled* — a fade pool.


---
## 7. Limitations, and what runs next

This cohort embeds a theory: informed flow = regular, taker-side, wallet-sized, medium-frequency.
Four known blind spots: **passive-side information** (maker fills were never scored), **episodic
traders** (crisis-only specialists fail the activity floors), **position reductions** (selling
before a dump is information our event definition ignores), and **delisting censoring** (dead
coins' final losses vanish from scoring — flagged, unfixed, flatters alt results).

**Declared next steps** (specs frozen before July data existed): July out-of-sample confirmation
of the basket and the alt-selection signal; the maker→taker transition detector (H4); crowd-
matched certification (H6); the delisting fix; a broad-universe existence pass; and the repaired
live paper harvester running the 43-slot basket.
